```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
#epsilon_r is relatieve permittiviteit van een materiaal (standaard 1 voor een vacuum)

import numpy as np

def vlakkeplaat(A, d, epsilon_r = 2.6): #de relatieve permittiviteit bij lucht of vacuum mag deze gelijkgesteld worden aan 1
    epsilon_0 = 8.854187817e-12  # permittiviteit van vacuum in F/m
    C = (epsilon_0 * epsilon_r * A) / d
    return C

#voor onze platencondensator
#gegeven

A_cap = 0.01      #m^2
d_cap = 0.002     #m

antwoord = vlakkeplaat(A_cap, d_cap, epsilon_r = 2.6)*1e12    #pF

# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.

A_array = np.array([A_cap])
d_array = np.array([d_cap])

def vlakkeplaat(A_array, d_array, epsilon_r = 2.6): #de relatieve permittiviteit bij papier zit tussen de 2.2 en 3
    epsilon_0 = 8.854187817e-12  # permittiviteit van vacuum in F/m
    C = epsilon_0 * epsilon_r * A / d
    print(C)
    return C

print(f"Berekende capaciteit plaatcondensator groep 53 is {antwoord:.1f} pF")

```

```python
# Plot hier je oorspronkelijke en je uiteindelijke kalibratiegrafieken, fit daar een bijpassende lijn door
# en zet ook de verwachte ontwerpgrafiek erbij

# De eerder geplotte kalibratiegrafiek nu met een curve_fit

massa = np.array([270.6, 277.2, 409.1, 1630.0, 686.3, 956.9, 289.6, 538.8, 828.4, 560.2, 679.7,1218])
arduino_waarden = np.array([48.7, 50.4, 69.1, 137.6, 98.8, 115.5, 52.3, 84.6, 108.6, 86.8, 98.1,126.5 ])    #pF

#curve_fit met een linear verband uit vlakkenplaat

# Fysisch model:
# Capaciteit van een vlakkeplaatcondensator:
# C = ε0 * εr * A / d

# De afstand d verandert door indrukking van de spons.
# Volgens de wet van Hooke:
# F = k * Δx  en  F = m*g
# Δx = (m*g)/k

# Nieuwe plaatafstand:
# d = d0 - Δx

def vlakkeplaat(A, d, epsilon_r = 2.6):                 #de relatieve permittiviteit bij papier zit tussen de 2.2 en 3

    epsilon_0 = 8.854187817e-12
    A = 0.01                                            #oppervlakte plaat
    g = 9.81                                            #gravitatieversnelling 
    d_0 = 1                                             #startafstand tussen de platen (met spons ertussen)  
    k = 200                                             #voorbeeld veerconstante van een spons

    d = (massa * g) / k                                 #wet van hook met veerconstante k
    C = epsilon_0 * epsilon_r * A / (d_0 - d)
    return C 

#vandaar dat de fit gedaan wordt met het volgende model
#onze x is gegeven (massa) en de fitwaarde c = -g(9.81) / k 

def functie(x, a, b,c):
    return a/(b+c*x)

popt, pcov = curve_fit(functie, massa, arduino_waarden)
a, b,c = popt
print(a,b,c)

massa_fit = np.linspace(min(massa), max(massa), 400)
fit_values = functie(massa_fit, a, b, c)

#curve_fit met een logaritmisch verband

def functie_2(x, a, b):
    return np.log(x)*a + b

popt_2, pcov_2 = curve_fit(functie_2, massa, arduino_waarden)
a, b = popt_2

massa_fit_2 = np.linspace(min(massa), max(massa), 400)
fit_values_2 = functie_2(massa_fit_2, a, b)

#toevoegen errorbars bij punten

#linear indruk spons duidelijk niet linear indrukken
#ontwerpplot

u_m= 10                            # u_m   = onzekerheid in massa (g)
u_ard= 0.1                         # u_ard = meetonzekerheid capaciteit (pF)

plt.figure(figsize=(8,5))
plt.plot(massa, arduino_waarden, 'k.', label = "verband massa en capaciteit")
plt.plot(massa_fit, fit_values, 'r--', linewidth=2, label=f"Fit: C = {a:.4f}·√m + {b:.2f}")
plt.plot(massa_fit_2, fit_values_2, 'k--', linewidth = 2, label = 'fit')
plt.xlabel("Massa (g)", fontsize=10)
plt.ylabel("Capaciteit (pF)", fontsize=10)

# errorbars tekenen
plt.errorbar(massa, arduino_waarden, xerr=u_m, yerr=u_ard, fmt='k.', ecolor='gray', elinewidth=1,capsize=3, label="Meetpunten + errorbars")

plt.title("Verband tussen massa en capaciteit", fontsize=14)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Opdracht 10: Vul hieronder de conclusie in.

De sensor is geschikt om verschillende gewichten te kunnen meten en dus te functioneren als een weegschaal. Door te itereren is dit experiment veel nauwkeuriger afgerond dan dat er eerst werd gedacht. Door het gebruik van papier in plaats van een spons als materiaal tussen de geleidende platen veranderde de capaciteit meer bij verschillende massa's. Door dit te vervangen door papier werd de afstand tussen de platen veel stabieler en kleiner, waardoor de capaciteitsverandering bij verschillende massa’s veel sterker en beter meetbaar werd. De sensor presteerde hierdoor aanzienlijk beter dan vooraf verwacht.

Er zit wel een significant verschil tussen de voorspelling van de ontwerpgrafiek en de uiteindelijke kalibratiegrafiek.De voorspelling gaf een logaritmisch verband, de gemeten punten liggen op een logaritmische lijn. De voorspelling ging uit van een omgekeerd evenredig verband, zoals volgt uit de vlakke‑plaat formule. In de praktijk bleken de meetpunten juist in een logaritmische kromme te liggen. De log-fit overeenkwam bijna exact met de data, terwijl de theoretische 1/x‑fit vrijwel niet aansloot.

Tot slot kan worden geconcludeerd dat de vlakke‑plaat formule onvoldoende nauwkeurig is voor deze specifieke sensor. De theoretische formule geeft het juiste type verband, maar de werkelijke meetdata volgt duidelijk een andere kromme. Voor deze sensor is een logaritmisch model daarom een betere benadering dan de ideale vlakke‑plaat vergelijking.

```python
# Plot hier je oorspronkelijke en je uiteindelijke kalibratiegrafieken, fit daar een bijpassende lijn door
# en zet ook de verwachte ontwerpgrafiek erbij

# De eerder geplotte kalibratiegrafiek nu met een curve_fit

massa = np.array([270.6, 277.2, 409.1, 1630.0, 686.3, 956.9, 289.6, 538.8, 828.4, 560.2, 679.7,1218])
arduino_waarden = np.array([48.7, 50.4, 69.1, 137.6, 98.8, 115.5, 52.3, 84.6, 108.6, 86.8, 98.1,126.5 ])    #pF

#curve_fit met een linear verband uit vlakkenplaat

# Fysisch model:
# Capaciteit van een vlakkeplaatcondensator:
# C = ε0 * εr * A / d

# De afstand d verandert door indrukking van de spons.
# Volgens de wet van Hooke:
# F = k * Δx  en  F = m*g
# Δx = (m*g)/k

# Nieuwe plaatafstand:
# d = d0 - Δx

def vlakkeplaat(A, d, epsilon_r = 2.6):                 #de relatieve permittiviteit bij papier zit tussen de 2.2 en 3

    epsilon_0 = 8.854187817e-12
    A = 0.01                                            #oppervlakte plaat
    g = 9.81                                            #gravitatieversnelling 
    d_0 = 1                                             #startafstand tussen de platen (met spons ertussen)  
    k = 200                                             #voorbeeld veerconstante van een spons

    d = (massa * g) / k                                 #wet van hook met veerconstante k
    C = epsilon_0 * epsilon_r * A / (d_0 - d)
    return C 

#vandaar dat de fit gedaan wordt met het volgende model
#onze x is gegeven (massa) en de fitwaarde c = -g(9.81) / k 

def functie(x, a, b,c):
    return a/(b+c*x)

popt, pcov = curve_fit(functie, massa, arduino_waarden)
a, b,c = popt
print(a,b,c)

massa_fit = np.linspace(min(massa), max(massa), 400)
fit_values = functie(massa_fit, a, b, c)

#curve_fit met een logaritmisch verband

def functie_2(x, a, b):
    return np.log(x)*a + b

popt_2, pcov_2 = curve_fit(functie_2, massa, arduino_waarden)
a, b = popt_2

massa_fit_2 = np.linspace(min(massa), max(massa), 400)
fit_values_2 = functie_2(massa_fit_2, a, b)

#toevoegen errorbars bij punten

#linear indruk spons duidelijk niet linear indrukken
#ontwerpplot

u_m= 10                            # u_m   = onzekerheid in massa (g)
u_ard= 0.1                         # u_ard = meetonzekerheid capaciteit (pF)

plt.figure(figsize=(8,5))
plt.plot(massa, arduino_waarden, 'k.', label = "verband massa en capaciteit")
plt.plot(massa_fit, fit_values, 'r--', linewidth=2, label=f"Fit: C = {a:.4f}·√m + {b:.2f}")
plt.plot(massa_fit_2, fit_values_2, 'k--', linewidth = 2, label = 'fit')
plt.xlabel("Massa (g)", fontsize=10)
plt.ylabel("Capaciteit (pF)", fontsize=10)

# errorbars tekenen
plt.errorbar(massa, arduino_waarden, xerr=u_m, yerr=u_ard, fmt='k.', ecolor='gray', elinewidth=1,capsize=3, label="Meetpunten + errorbars")

plt.title("Verband tussen massa en capaciteit", fontsize=14)
plt.grid(True)
plt.tight_layout()
plt.show()
```
