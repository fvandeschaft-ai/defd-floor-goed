```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
#epsilon_r is relatieve permittiviteit van een materiaal (standaard 1 voor een vacuum)

import numpy as np

def vlakkeplaat(A, d, epsilon_r = 2.6): #de relatieve permittiviteit bij lucht of vacuum mag deze gelijkgesteld worden aan 1
    epsilon_0 = 8.854187817e-12  # permittiviteit van vacuum in F/m
    C = (epsilon_0 * epsilon_r * A) / d
    return C

#voor onze platencondensator
#gegeven

A_cap = 0.01      #m^2
d_cap = 0.002     #m

antwoord = vlakkeplaat(A_cap, d_cap, epsilon_r = 2.6)*1e12    #pF

# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.

A_array = np.array([A_cap])
d_array = np.array([d_cap])

def vlakkeplaat(A_array, d_array, epsilon_r = 2.6): #de relatieve permittiviteit bij papier zit tussen de 2.2 en 3
    epsilon_0 = 8.854187817e-12  # permittiviteit van vacuum in F/m
    C = epsilon_0 * epsilon_r * A / d
    print(C)
    return C

print(f"Berekende capaciteit plaatcondensator groep 53 is {antwoord:.1f} pF")

```

```python
# Plot hier de verwachte ontwerpgrafiek van jullie sensoren.
# Het berekenen van de capaciteit volgens de gemeten spanning

C = np.array([]) *1e12  #gemeten capaciteit (pF)

def capaciteit(V_out, R, omega, V_in):
   C = V_out/(R*omega*V_in)
   return C

V_out = np.array([332, 531, 562, 603, 1186]) / 1000      #spanning (V)
R = 1000                                #ohm
omega = 628318                          #radialen per seconde
V_in = 8                                #volt

C = capaciteit(V_out, R, omega, V_in)*1e12          #pF
C_gem = np.mean(C)
#Met de capaciteit wordt de d (afstand tussen de platen) berekend

A = 0.01              # oppervlak plaat (m)
epsilon_r = 1.05      # relatieve permittiviteit van de spons

d = np.array([])   #de resultaten, afstand tussen platen (m), wordt opgeslagen in deze array

def spons_indruk(A, C, epsilon_r):
    epsilon_0 = 8.854187817e-12   
    d = epsilon_0 * epsilon_r * A / C
    return d

print(f"De minimale capaciteit (pF): {C[0]:.1f}")
print(f"De maximale capaciteit (pF): {C[4]:.1f}")
print(f"De gemiddelde capaciteit is {C_gem:.1f}")
```

```python
#Verwachtingen
#In deze cel laten we de verwachting zien van het verband is van de afstand tussen de platen en de capaciteit
#we verwachten een omgekeerd evenredig verband 

import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

C = np.array([83, 120, 140, 219])

def verwachting(C):
    d = np.sqrt(C)
    return d

d = verwachting(C) 

def model(C, k):
    return k * np.sqrt(C)

popt, pcov = curve_fit(model, C, d)
k_fit = popt[0]

C_fit = np.linspace(min(C), max(C), 100)
d_fit = model(C_fit, k_fit)

plt.figure(figsize=(8,5))
plt.plot(C, d, 'k.', label = "verband (1/C)")
plt.plot(C_fit, d_fit, 'r--', label = "fit" )
plt.xlabel("Capaciteit (pF)")
plt.ylabel("Afstand (m)")
plt.legend()
plt.title("Wortelverband tussen C en d")
plt.grid()
plt.show()
```
